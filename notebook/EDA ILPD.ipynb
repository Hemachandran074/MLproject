**Installing the Libraries**

```python
import pandas as pd
import numpy as nu
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

**Import the Dataset**

```python
df = pd.read_csv('data\indian_liver_patient.csv')
```

```python
df.head()
```

**Dataset Information**

1. Age : Age of the patient --> Numerical
2. Gender : Gender of the patient --> (Male. Female)
3. Total_Bilirubin : Total Bilirubin in Mg/dL --> Numerical
4. Direct_Bilirubin : Conjucated Bilirubin in Mg/dL --> Numerical
5. Alkaline_Phosphotase : ALP in IU/L --> Numerical
6. Alamine_Aminotransferase : ALT in IU/L --> Numerical
7. Aspartate_Aminotransferase : AST in IU/L --> Numerical
8. Total_Protiens : Total Proteins g/dL --> Numerical
9. Albumin : Albumin in g/dL --> Numerical
10. Albumin_and_Globulin_Ratio : A/G ratio --> Numerical
11. Dataset : Label (patient has liver disease or not) --> (0 and 1)

**Data Visualize**

```python
df.Gender.value_counts().plot(kind="pie",figsize=(6, 6), autopct='%.2f', title="Gender distribution")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.scatterplot(data=df, x='Total_Bilirubin',y='Direct_Bilirubin',hue='Gender',ax=axes[0])
sns.scatterplot(data=df, x='Total_Bilirubin',y='Direct_Bilirubin',hue='Dataset',ax=axes[1])
plt.show()
```
